## Q35 

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
# output:[[1,99],[3,4]]
```

```python
# the copy is shallowcopy()
# shallow copy():It copies top level structure,nested objects are still shared
# so, changes list 'b' will effect in list 'a'
```

## Q36

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
# output:[[1,2],[3,4]]
```

```python
# the copy is deepcopy()
# deepcopy():creates a completely independent copy, including the inner lists
# so,the changes in 'b' list will not effect list 'a'
```
