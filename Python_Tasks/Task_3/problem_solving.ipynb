### Q1. Find the Frequency of Each Word

<!-- 
1. take input sentence and split into words  
2. create an empty frequency dictionary  
3. iterate each word through for loop  
4. check whether the word exists in the dictionary  
5. increase count if it exists, otherwise assign count as 1  
6. return and print frequency dictionary -->

<!-- START
DEFINE function word_f(sentence)
    SPLIT sentence into words
    CREATE an empty dictionary frequency

    FOR each word in words:
        IF word exists in frequency:
            INCREASE its count by 1
        ELSE:
            SET its count to 1

    RETURN frequency

INPUT sentence
CALL word_f(sentence)
PRINT the result
STOP -->

```python
def word_f(sentence):
    words = sentence.split()
    frequency = {}

    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1

    return frequency
sentence = input("Enter a sentence: ")
result = word_f(sentence)
print(result)
```

### Q7. Find the Second Highest Distinct Number in a List


<!-- 1. take input a list of numbers  
2. remove duplicate values using set and convert back to list  
3. sort the unique numbers in ascending order  
4. check whether the list contains at least two unique numbers  
5. return None if fewer than two exist, otherwise select the second-largest number  
6. print the returned result -->

```python
def second_highest(numbers):
    unique = list(set(numbers))
    unique.sort()
    if len(unique) < 2:
        return None
    return unique[-2]
numbers = [10, 20, 40, 30, 40, 20]
print(second_highest(numbers))
```

### Q10. Find Two Numbers Whose Sum Equals a Target


<!-- 1. take input a list of numbers and target value  
2. iterate list elements using outer for loop  
3. iterate the remaining elements using inner for loop  
4. calculate the sum of the two selected elements  
5. if the sum equals the target, return the pair as a list  
6. if no pair is found, return and print an empty list -->

```python
def find_pair(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [nums[i], nums[j]]
    return []
nums = [2, 7, 11, 15]
target = 9
p=find_pair(nums, target)
print(p)
```

### Q13. Rotate a List to the Right by K Positions


<!-- 1. take input a list of numbers and rotation value k  
2. check whether the list is empty and return it if empty  
3. calculate effective rotation using k modulo list length  
4. select the last k elements of the list  
5. combine the last k elements with the remaining elements, or return the original list if k is zero  
6. return and print the rotated list -->

```python
def rotate_right(nums, k):
    if len(nums) == 0:
        return nums
    k = k % len(nums)
    return nums[-k:] + nums[:-k] if k != 0 else nums
nums = [1, 2, 3, 4, 5]
k = 2
u=rotate_right(nums, k)
print(u)
```

### Q14. Find the Maximum Sum of Two Consecutive Elements


<!-- 1. take input a list containing at least two numbers  
2. assume the sum of the first two elements as maximum  
3. iterate consecutive pairs through for loop  
4. calculate the sum of each consecutive pair  
5. compare each sum with maximum and update maximum if greater  
6. return and print the maximum sum -->

```python
def max_consecutive_sum(nums):
    maximum = nums[0] + nums[1]
    for i in range(1, len(nums) - 1):
        total = nums[i] + nums[i + 1]
        if total > maximum:
            maximum = total
    return maximum
nums = [4, 7, 2, 9, 5]
e=max_consecutive_sum(nums)
print(e)
```

### Q15. Find All Pairs of Numbers Whose Sum Equals a Target


<!-- 1. take input a list of numbers and target value  
2. create an empty list to store valid pairs  
3. select pairs using nested for loops with different indices  
4. check whether the sum of each pair equals the target  
5. sort each valid pair and append it only if it is not already present  
6. sort all pairs, return the result, and print it -->

```python
def all_pairs(nums, target):
    pairs = []
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pair = sorted([nums[i], nums[j]])
                if pair not in pairs:
                    pairs.append(pair)
    pairs.sort()
    return pairs
nums = [2, 4, 3, 5, 7, 8, 1]
target = 9
a=all_pairs(nums, target)
print(a)
```
